```python
import numpy as np
```

```python
print(np.pad(np.zeros((8, 8), dtype=int), 1, constant_values=1))
```

```python
print( np.tile([[0, 1], [1, 0]], (4, 4)))
```

```python
print(np.fromfunction(lambda i, j: (i > j) * (i - j), (5, 5), dtype=int))
```

```python
print((lambda x: x - x.mean(axis=1, keepdims=True))(np.random.rand(5, 5)))
```

```python
print((lambda a: a[np.arange(10), np.abs(a - 0.5).argmin(axis=1)])(np.random.rand(10, 3)))
```

```python
print((lambda a: a.sum(axis=1) / a.max(axis=0))(np.random.rand(6, 6)))
```

```python
print(np.array([6, 2, 0, 3, 0, 0, 5, 7, 0])[1:][np.array([6, 2, 0, 3, 0, 0, 5, 7, 0])[:-1] == 0].max())
```

```python
print(np.ones(10) + np.bincount(np.array([0, 1, 2, 3, 5, 5, 5, 8]), minlength=10))
```

```python
a = np.array([[1, 4.9, 2.4], [7.1, 7, 9.8], [2, 4.6, 1.2]]); b = np.array([3, 5, 10, 0]); print(b[np.abs(a[..., None] - b).argmin(axis=-1)])
```

```python
a = np.array([1, 4.9, 2.4, 7.1, 7, 9.8, 2, -5, 4.6, 1.2]); print(np.convolve(a, np.ones(3)/3, mode='valid'))
```
